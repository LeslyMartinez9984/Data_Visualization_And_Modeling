```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
df = pd.read_csv('SteamGames.csv')
df.head()
print(df.columns)
```

#Summary
This dataset has information about games that are available in Steam. It includes data like their genre, price, developer, ect. The target of this dataset is to show the review score, which would show the overall review rating of the game. The goal is see what features seem to have the a strong effect on the game's review score.

```python
#Positive Reviews VS Review Score
plt.scatter(df['PositiveReview'], df['ReviewScore'])
plt.xlabel('Positive Reviews')
plt.ylabel('Review Score')
plt.title('Positive Reviews vs Review Score')
plt.show()
```

The scatterplot is comparing the number of positive reviews that a game has recieved with its overall review score. Most games with very high review scores appear to have received a big number of positive reviews, suggesting that positive feedback may contribute to stronger overall ratings. However, some games with fewer positive reviews also achieved high review scores.

```python
#Negative Reviews VS Review Score
plt.scatter(df['NegativeReview'], df['ReviewScore'])
plt.xlabel('Negative Reviews')
plt.ylabel('Review Score')
plt.title('Negative Reviews vs Review Score')
plt.show()
```

The scatterplot is comparing the number of negative reviews that a game has recieved with its overall review score. Most games with very high review scores appear to have received a lot of negative reviews, suggesting that nagative feedback may not contribute to stronge overall rating. Also, games with a high review score seem to have a lot of nagative reviews.

```python
#Price VS Review Score
plt.scatter(df['price'], df['ReviewScore'], alpha=0.5)
plt.xscale('log')
plt.xlabel('Price')
plt.ylabel('Review Score')
plt.title('Price vs Review Score')
plt.show()
```

The graph shows the relationship between game prices and then their review scores. The relationship between them is shown to be pretty weak there's no correlation between the prices and the review score.

```python
#Type VS Reveiw Score
df.boxplot(column='ReviewScore', by='Type', rot=45)
plt.title('Type vs Review Score')
plt.suptitle(' ')
plt.xlabel('Type')
plt.ylabel('Review Score')
plt.show()
```

The boxplot is comparing the game types with their review scores. The dlc and games seem to average higher review scores than the other types of games. Then hardware and view seem to have lower review score to the rest. Then music rangers from high to low.

```python
#Rank VS Review Score
plt.scatter(df['Rank'], df['ReviewScore'])
plt.xlabel('Rank')
plt.ylabel('Review Score')
plt.title('Rank vs Review Score')
plt.show()
```

The scatterplot is comparing the game's rank with their review score. There isn't much effect from the rank since it seems that their review score is all over the place no matter the rank.

#Conclusion
The dataset does show hat the positive reviews show a strong positive relationship with review scores, while negative reviews don't show as much of a difference. Also, features such as price appear to have much less influence on how players rate games.  Overall, the data indicates that player satisfaction is more closely related to review activity and community response than to a game's cost.
