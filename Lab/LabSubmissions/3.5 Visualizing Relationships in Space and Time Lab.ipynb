```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('GlobalWeatherRepository.csv')
df.head()
print(df.columns)
```

#Summary
This dataset has global weather information from different cities and countries. This includes the location, dates, temperature, humidity, wind speed, etc. The purpose of the data is to analyse how weather conditions change across different locations and time.

```python
#Temperature Over Time
df['last_updated'] = pd.to_datetime(df['last_updated'])
daily_avg = df.groupby(df['last_updated'].dt.date)['temperature_celsius'].mean()
plt.figure(figsize=(12,5))
plt.plot(daily_avg.index, daily_avg.values)
plt.xlabel('Date')
plt.ylabel('Temperature C')
plt.title('Daily Average Temperature Over Time')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The data shows how the average daily temperature has changed throughout a period of time. It also, shows the temperature fluctuates at certain times of the year, with it also having random spikes of cold.

```python
#Average Temperature By Country
avg_temp = df.groupby('country')['temperature_celsius'].mean()
avg_temp = avg_temp.sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
avg_temp.plot(kind='bar')
plt.xlabel('Country')
plt.ylabel('Average Temperatur (C)')
plt.title('Top 10 Countries By Average Temperature')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The bar chart helps show the different average temperatures in different countries. The temperature between these countries does vary a bit but looking at like Oman and after seem to have pretty similar temperature averages. However, Saudi Arabien does seem to have higher temperatures than the other countries.

```python
#Humidity Over Time
df['last_updated'] = pd.to_datetime(df['last_updated'])
df = df.sort_values('last_updated')
countries = df['country'].dropna().unique()[:3]
plt.figure(figsize=(12,5))
for country in countries:
    temp = df[df['country'] == country]
plt.plot(temp['last_updated'],
         temp['humidity'],
         label=country)
plt.xlabel('Date')
plt.ylabel('Humidity %')
plt.title('Humidity Over Time by Country')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The graph compares the humidity levels over a time period across multiple locations. It also, shows the variation between different times of year with a specific spike in early 2026.
