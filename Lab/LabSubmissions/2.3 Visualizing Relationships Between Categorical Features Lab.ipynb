```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('titanic_passengers.csv')
df.head()
df.columns
```

```python
#Survival By Sex
survival_by_sex = df.groupby('Sex')['Survived'].mean()
survival_by_sex.plot(kind='bar')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Sex')
plt.show()
print(survival_by_sex)
```

Female passengers had a survival rate of about 74%, while male passengers had a survival rate of about 19%. This shows that women were significantly more likely to survive the Titanic disaster. It does help support the evacuation process used back then to have women and children evacuate first.

```python
#Survival By Class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
survival_by_class.plot(kind='bar')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Passenger Class')
plt.show()
print(survival_by_class)
```

The first class passengers had the highest survival rate at about 63%, while the third class passangers had the lowest with only about a 24%. Then the second class passengers were in between them with a 47% survival rate. This shows that social class did play a role in the survival chances during the Titanic sinking.

```python
#Survival By Age
df['Age_Group'] = 'Adult'
df.loc[df['Age'] < 18, 'Age_Group'] = 'Child'
df.loc[df['Age'] >= 60, 'Age_Group'] = 'Senior'
age_survival = df.groupby('Age_Group')['Survived'].mean()
age_survival.plot(kind='bar')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Age')
plt.show()
print(age_survival)
```

Children had the highest survival rate of about 54%, while adults had a lower survival rate around 36%. Then senior passengers had the lowest survival rate of around 27%. The results show that the younger passengers were more likely to survive the Titanic sinking, with the procedure to ususally evacuate children first.

```python
#Survival By Fare
df['Fare_Group'] = 'Low Fare'
df.loc[df['Fare'] > 50, 'Fare_Group'] = 'High Fare'
fare_survival = df.groupby('Fare_Group')['Survived'].mean()
fare_survival.plot(kind='bar')
plt.ylabel('Survival Rate')
plt.title('Survival Rate By Fare')
plt.show()
print(fare_survival)
```

Passengers who had paid higher fares had higher survival rate of about 68%. Then those who had paid a lower fare had a way lower survival rate of about 32%. This shows that the welathier passengers had a better chance of survival than those who didn't pay for a higher fare.

```python
#Survival By Embarked Port
embarked_survival = df.groupby('Embarked')['Survived'].mean()
embarked_survival.plot(kind='bar')
plt.ylabel('Survival Rate')
plt.title('Survival Rate By Embarked Port')
plt.show()
print(embarked_survival)
```

Passengers who embarked from Cherbourg had the highest survival rate at about 55%,while those who embarked at Queenstown had a survival rate of about 39%. Then lastly those who embarked at Southampton had the lowest survival rate at about 34%. This shows that whiever port the passengers departed from does correlate to their survival rate because of their social classes.
