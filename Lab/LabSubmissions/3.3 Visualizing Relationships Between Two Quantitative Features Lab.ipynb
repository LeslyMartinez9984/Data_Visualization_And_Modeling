```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
df = pd.read_csv('Employee_productivity.csv')
df.head()
df.columns
```

```python
#Years At Company VS Productivity
plt.scatter(df['YearsAtCompany'], df['ProductivityScore'])
plt.xlabel('Years At Company')
plt.ylabel('Productivity Score')
plt.title('Years At Company vs Productivity')
plt.show()
x = df[['YearsAtCompany']]
y = df['ProductivityScore']
model = LinearRegression()
model.fit(x, y)
predictions = model.predict(x)
print("R-Squared:", r2_score(y, predictions))
```

The scatterplot shows a small relationship between the number of years at the compony and productivity. The R-squared value is low, which means that years at the company doesn't predict employee productivity. So, whether new or experienced employee it doesn't make a big impact in the productivity score.

```python
#Hours Worked VS Productivity
plt.scatter(df['HoursWorked'], df['ProductivityScore'])
plt.xlabel('Hours Worked')
plt.ylabel('Productivity Score')
plt.title('Hours Worked vs Productivity')
plt.show()
x = df[['HoursWorked']]
y = df['ProductivityScore']
model = LinearRegression()
model.fit(x, y)
predictions = model.predict(x)
print("R-Squared:", r2_score(y, predictions))
```

The scatterplot shows no real relationship between the hours worked and the productivity that was made. The R-squared value is closer to 0, which shows that the hours worked doesn't have a big impact in the productivity score. Even if employees work more hours it doesn't correlate with the productivity.

```python
#Department VS Productivity
df.boxplot(column='ProductivityScore', by='Department')
plt.xlabel('Department')
plt.ylabel('Productivity Score')
plt.title('Department vs Productivity')
plt.suptitle('')
plt.show()
```

The boxplot shows that the productivity score are pretty similar in all deparments. The median of the productivity score for all them are pretty close especially the HR, Marketing, and Sales. This shows that the department doesn't have a big impact on the productivity scores.
