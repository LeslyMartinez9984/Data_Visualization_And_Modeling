#Questions
1) What are the top 10 highest rated anime?

```python
#Highest Rated Anime
top_rated = df.sort_values(by='score', ascending=False)
top_rated[['title', 'score']].head(10)
```

The highest rated anime in the dataset is “Sousou no Frieren” with a score of 9.27. Other highly rated anime includec “Fullmetal Alchemist: Brotherhood,” “Steins;Gate,” and “Hunter x Hunter (2011).” Many of the top rated anime are well known series with large fanbases and strong opinionated reviews.

```python
#Most Common Genres
genres = df['genres'].dropna().str.split(',')
all_genres = genres.explode()
all_genres = all_genres.str.strip()
all_genres.value_counts().head(10)
```

The most common genre in the dataset is Comedy, appearing over 3,000 times. Other frequently watched genres include Fantasy, Drama, and Slide of Life. This shows that more lighthearted imaginative genres are more popular when it comes to anime.

```python
#Average Anime Score
df['score'].mean()
```

The average anime score in the dataset is about 6.43. This shows that most of the anime receive a little more positive ratings, with a few very high rated titles that are increasing the overall average.

```python
#Studios With Most Anime Produced
df['studios'].value_counts().head(10)
```

The studio that appears most frequently in the dataset is Toei Animation, followed by Sunrise and J.C.Staff. These studios have produced a large number of anime titles, which show that they are some of the most active and influential production companies in the industry. This shows that a small number of major studios are responsible for a large portion of anime that has been created.

```python
#Episodes VS Score
df[['episodes', 'score']].corr()
```

The comparison between the number of episodes and their anime score is about 0.095, which indicates a very weak positive relationship. This suggests that there is litte to nothing connecting how long an anime is and how highly rated it is. Both short and long anime could receive a high or low score, not depending on the amount of episodes they have.
