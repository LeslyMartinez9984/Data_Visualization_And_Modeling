```python
import pandas as pd
import matplotlib.pyplot as plt

#Loading DataSet
df = pd.read_csv('spotify-2023.csv', encoding='ISO-8859-1')

#View The First Rows
df.head()

#Visualizing BPM Distribution
plt.hist(df['bpm'], bins=20)
plt.xlabel('BPM')
plt.ylabel('Number of Songs')
plt.title('Distribution of BPM')
plt.show()
print(df['bpm'].describe())
```

The BPM values in the dataset range from 65 to 206 BPM, with an average BPM of about 123. Most Songs fall between 100 and 140 BPM, as shown by the 25th and 75th percentiles. It suggest that popular songs tend to have moderately fast tempos that have energy to them.

```python
#Visualizing Key Distribution
df['key'].value_counts().plot(kind='bar')
plt.xlabel('Key')
plt.ylabel('Count')
plt.title('Distribution of Musical Keys')
plt.show()
print(df['key'].value_counts())
```

The most common musical keys are C#, G, G#, and F, with C# being the most frequent. Keys D#, A#, and E seem to sho up less often. This shows that while songs are spread across many keys, a few of them are more popular as shown by the dataset.

```python
#Visualizing Mode Distribution
df['mode'].value_counts().plot(kind='bar')
plt.xlabel('Mode')
plt.ylabel('Count')
plt.title('Distribution of Mode')
plt.show()
print(df['mode'].value_counts())
```

Most songs in the dataset are in Major, with the mode of 550 songs. There are fewer songs that are in Minor with the mode being 403. This shows that popular music is favoring Major, which correlate to more of a brighter sound.

```python
#Danceability Distribution
plt.hist(df['danceability_%'], bins=20)
plt.xlabel('Danceability %')
plt.ylabel('Number of Songs')
plt.title('Distribution of Danceability')
plt.show()
print(df['danceability_%'].describe())
```

The danceability values range from 23% to 96%, with an average of about 67%. Most songs fall between 57% and 78% danceability, showing that popular songs are in general easier to dance to and have a strong  rhythmuc composition.

```python
#Energy Distribution
plt.hist(df['energy_%'], bins=20)
plt.xlabel('Energy %')
plt.ylabel('Number of Songs')
plt.title('Distribution of Energy')
plt.show()
print(df['energy_%'].describe())
```

The energy values ranege from 9% to 97%, with an avergae energy level of about 64%. Most songs fall between 53% and 77%, which seems to represents that popular songs tend to have moderate to high energy levels.

```python
#Speechiness Distribution
plt.hist(df['speechiness_%'], bins=20)
plt.xlabel('Speechiness %')
plt.ylabel('Number of Songs')
plt.title('Distribution of Speechiness')
plt.show()
print(df['speechiness_%'].describe())
```

The speechiness values range from 2% to 64%, with an average of about 10%. Most songs fall between 4% and 11%, which shows that popular songs don't contain much on spoken content.

```python
#Guranteed Smash Hit Criteria
hit_song = df[
    (df['bpm'] >= 115) &
    (df['bpm'] <= 130) &
    (df['key'].isin(['C#', 'G'])) &
    (df['mode'] == 'Major') &
    (df['danceability_%'] >= 65) &
    (df['danceability_%'] <= 80) &
    (df['energy_%'] >= 65) &
    (df['energy_%'] <= 85) &
    (df['speechiness_%'] < 15) &
    (df['acousticness_%'] < 30)
]
hit_song[['track_name', 'artist(s)_name', 'streams']].head(10)
```

Many songs have matched with this specific criteria, which can be considered as smash hit. Some examples are “Cold Heart - PNAU Remix” by Dua Lipa and Elton John, “We Found Love” by Rihanna and Calvin Harris, and “Pepas” by Farruko. Many of these songs were highly successful and had very high stream counts, with a lot of them reaching over 1 billion on Spotify. For example, “We Found Love” by Rihanna and Calvin Harris had over 1.2 billion streams, showing that songs matching these characteristics can achieve major commercial success. This shows that this type of criteria is common among popular songs.

#Conclusion
Based on the visualizations and statiscal analysis of the Spotify dataset, popular songs tend to have several common characteristics. Most of the songs have a BPM between 100BPM and 140BPM, moderate to high danceability, and medium to high energy levels. The most common musical key was C#, and in Major key appeared more frequently. Also, most songs had relatively low speechiness and lower acoustics, showing that mainstream hits generally are more melodic and heavily produced. After applying these characteristics to create a “smash hit” formula, many of these songs had extremely high stream counts, with some exceeding one billion streams on Spotify. This suggests that these musical characteristics are commonly associated with successful popular songs, although they cannot completely guarantee a hit.
