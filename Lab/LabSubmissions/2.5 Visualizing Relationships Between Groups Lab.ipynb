```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('loan_data.csv')
df.head()
df.columns
```

```python
#Age VS Loan
df.boxplot(column='person_age', by='loan_status')
plt.title('Age vs Loan Status')
plt.suptitle('')
plt.xlabel('Loan Status')
plt.ylabel('Age')
plt.show()
```

The boxplot shows that the applicants with approved loans tend to be younger than the applicants who get rejected. The jected applicants seem to have a higher median age and a wider age distribution. This shows that age can have influence on whether or not an applicant will be approved for a loan, which would be an age related bias.

```python
#Sex VS Loan
sex_survival = pd.crosstab(df['person_gender'], df['loan_status'])
sex_survival.plot(kind='bar')
plt.title('Gender vs Loan Status')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
print(sex_survival)
```

The number of aproved and rejected loans depending on sex is showing to be pretty close. Even though overall there are more male applicants, both show similar rates of approvals and rejections. The sex of someone doesn't seem to strongly influence the approval decision for a loan.

```python
#Income VS Loan
df.boxplot(column='person_income', by='loan_status')
plt.title('Income vs Loan Status')
plt.suptitle('')
plt.xlabel('Loan Status')
plt.ylabel('Income')
plt.show()
```

The boxplot shows that the applicants with higher incomes seem to be getting  approved more for loans than those who report a lower one. The median income for approved loans is higher, which shows that depending on the applicants income can play a factor in whether they get approved or not for the loan. 

```python
#Credit Score VS Loan
df.boxplot(column='credit_score', by='loan_status')
plt.title('Credit Score vs Loan Status')
plt.suptitle('')
plt.xlabel('Loan Status')
plt.ylabel('Credit Score')
plt.show()
```

The box plot is showing that applicants with a higher credit score tend to be approved a little bit more. The median on those who get approved is higher, which shows that depending on your credit score could have an influence on whether an applicant is approved.

```python
#Loan Percent Income VS Loan
df.boxplot(column='loan_percent_income',by='loan_status')
plt.title('Loan Percent Income vs Loan Status')
plt.suptitle('')
plt.xlabel('Loan Status')
plt.ylabel('Loan Percent Income')
plt.show()
```

The box plot shows that applicants that are being approved for loans have a lower loan percent income than those who are getting rejected. The median for the applicannts that are being rejected is higher. This shows that the applicants loan percent income does play a big factor in whether or not the application gets approved.

```python
#Previous Loan Defaults VS Loan
defaults = pd.crosstab(df['previous_loan_defaults_on_file'], df['loan_status'])
defaults.plot(kind='bar')
plt.title('Previous Defaults vs Loan Status')
plt.xlabel('Previous Defaults')
plt.ylabel('Count')
plt.show()
print(defaults)
```

It shows that applicants with previous loan defaults show a noticeably higher loan rejections compared to those without defaults. This suggests that prior default history is a strong factor in the deciding whether or not to approved the loan.

```python
#Home Ownership VS Loan
ownership = pd.crosstab(df['person_home_ownership'], df['loan_status'])
ownership.plot(kind='bar')
plt.title('Home Ownership vs Loan Status')
plt.xlabel('Home Ownership')
plt.ylabel('Count')
plt.show()
```

This shows that applicants who own homes are more likely to get approved for loans than those who rent or have a mortgage. This suggests that depending on whether you own or have a mortgage really plays a big factor on getting approved for a loan.

#Conclusion
Based on all data, the most influential factors in loan approval decisions seem to be income, credit score, loan-to-income ratio, and previous loan defaults. These factors are showing big differences between approved and rejected applicants. However, age and gender do not show strong or consistent differences in approval outcomes. While age shows some variation, it is not strong enough on its own to confirm a bias. Overall, the data seems to rely primarily on financial risk indicators rather than demographic features.
