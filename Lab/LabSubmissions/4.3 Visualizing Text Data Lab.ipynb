```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
df = pd.read_csv('Rihanna.csv')
df.head()
print(df.columns)
```

#Summary
This dataset has song lyrics from the artist Rihanna's songs, and information about each of the songs titles, albums, and other information. It is used to show the patterns through the song lyrics and the different themes between the songs. 

```python
#Most Common Lyrics
text = " ".join(df['Lyric'].astype(str))
lyrics = text.lower().split()
common_lyrics = Counter(lyrics).most_common(10)
lyrics_df = pd.DataFrame(common_lyrics, columns=['Lyric', 'Count'])
plt.bar(lyrics_df['Lyric'], lyrics_df['Count'])
plt.xlabel('Lyrics')
plt.ylabel('Frequency')
plt.title('Most Common Lyrics in Rihanna Lyrics')
plt.xticks(rotation=45)
plt.show()
```

The graph shows what lyrics are more commonly used in her songs. It seems to be really common words in general that refer to oneself as well as just simple words like it or and.

```python
#Songs Per Album
album_counts = df['Album'].value_counts()
plt.figure(figsize=(12,6))
album_counts.plot(kind='bar')
plt.xlabel('Album')
plt.ylabel('Number of Songs')
plt.title('Number of Songs Per Album')
plt.xticks(rotation=90)
plt.show()
```

The bar chart is showing the number of songs in each Rihanna album from the dataset. It does show that eventually with each album realease less songs were being implemented into the albums.

```python
#Songs Released By Year
year_counts = df['Year'].value_counts().sort_index()
plt.figure(figsize=(10,5))
year_counts.plot(kind='line', marker='o')
plt.xlabel('Year')
plt.ylabel('Number of Songs')
plt.title('Songs Released by Year')
plt.show()
```

The graph does show how different the amount of songs that Rihanna has realeased throughout the years. Then 2010 - 2012 having a high spike as well as 2016 spiking up again. However it does show that her song releases have dwindled in recent years.

#Conclusion
This dataset provided details about Rihanna's music by showing common used song lyrics, albums, and song release years. The most common words found in the lyrics were simple and frequently used words such as it, and, and words that refer to oneself. This shows that many of Rihanna's songs use conversational language. Then the albums graph showed differences in the number of songs included on each album. Earlier albums generally contained more songs, while some later albums included fewer tracks. The release year graph showed song releases increased significantly between 2010 and 2012 and experienced another noticeable increase around 2016. However, the number of songs released in more recent years has decreased. Overall, the dataset showed patterns in Rihanna's lyrical content, album composition, and music production over time.
